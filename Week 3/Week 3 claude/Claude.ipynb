```python
import pandas as pd
from numpy import mean, absolute
from sklearn.model_selection import RepeatedKFold, cross_val_score
from sklearn.linear_model import LinearRegression
```

```python
insurance = pd.read_csv("insurance.csv")
insurance = pd.get_dummies(insurance, dtype=int)
insurance = insurance.dropna()   # remove missing values (there are none, but it's good practice)
insurance.head()
```

```python
y = insurance['charges']                 # target
X = insurance.drop('charges', axis=1)    # all other columns are features
```

```python
cv = RepeatedKFold(n_splits=5, random_state=420)

lm = LinearRegression()
lmscores = cross_val_score(lm, X, y,
                           scoring='neg_mean_absolute_error',
                           cv=cv)

lmMAE = mean(absolute(lmscores))
print('The average prediction error with all features is: %.0f' % lmMAE)
```

```python
X_nosex = X.drop(['sex_female', 'sex_male'], axis=1)

lmscores_nosex = cross_val_score(lm, X_nosex, y,
                                 scoring='neg_mean_absolute_error',
                                 cv=cv)

lmMAE_nosex = mean(absolute(lmscores_nosex))
print('The average prediction error without sex is: %.0f' % lmMAE_nosex)
```
