```python
import pandas as pd
from numpy import mean, absolute
from sklearn.model_selection import RepeatedKFold, cross_val_score
from sklearn.linear_model import LogisticRegression   

# Load and prepare the data
insurance = pd.read_csv("insurance2.csv")
insurance = pd.get_dummies(insurance, dtype=int)
insurance = insurance.dropna()

# Target and features
y = insurance['death']
X = insurance.drop('death', axis=1)

# Prediction with k-fold cross-validation
cv = RepeatedKFold(n_splits=5, random_state=420)
logr = LogisticRegression(max_iter=10000)
logscores = cross_val_score(logr, X, y, scoring='accuracy', cv=cv)

print('Accuracy: %.3f' % mean(logscores))

# Baseline: how often did people not die?
print(y.value_counts(normalize=True))
```
