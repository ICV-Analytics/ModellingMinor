```python
import pandas as pd 

insurance = pd.read_csv("insurance.csv")
insurance = pd.get_dummies(insurance, dtype=int)
insurance.head() 
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import KFold, cross_val_predict
```

```python
y = insurance["charges"]
X_all = insurance.drop(columns=["charges"])
```

```python
kf = KFold(n_splits=5, shuffle=True, random_state=42)
```

```python
model = LinearRegression()
predicted_all = cross_val_predict(model, X_all, y, cv=kf)

mae_all = abs(y - predicted_all).sum() / len(y)
print("MAE with all features:", mae_all)
```

```python
X_no_sex = insurance.drop(columns=["charges", "sex_female", "sex_male"])

predicted_no_sex = cross_val_predict(model, X_no_sex, y, cv=kf)

mae_no_sex = abs(y - predicted_no_sex).sum() / len(y)
print("MAE without sex:", mae_no_sex)
```
